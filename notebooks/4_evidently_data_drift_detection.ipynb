## Topic: Data Drift Detection Demo using Evidently on BostonHousing Dataset

```python
## import Require libraries

import pandas as pd
import numpy as np

from sklearn import datasets

from evidently.report import Report
from evidently.metric_preset import DataDriftPreset
```

```python
# load the BostonHousing data from github
data_url = "https://raw.githubusercontent.com/selva86/datasets/master/BostonHousing.csv"

# convert to dataframe
boston_frame = pd.read_csv(data_url)

boston_frame.head()
```

```python
# info
boston_frame.info()
```

```python
# shape
boston_frame.shape
```

```python
# Training set 
reference_data = boston_frame.iloc[:200]

# testing set
current_data = boston_frame.iloc[200:]
```

```python
reference_data
```

```python
data_drift_report = Report(
    metrics=[
        DataDriftPreset()
    ]
)
```

```python
data_drift_report.run(
    reference_data=reference_data,
    current_data=current_data,
    column_mapping=None
)
# Save the report as HTML
data_drift_report.save_html("data_drift_report.html")
```
